```python
import pandas as pd
import numpy as np
import tensorflow as tf
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Full_punch.csv')
window_size = 100
features_list = []
```

```python
for punch_type in data['Type'].unique():
    subset = data[data['Type'] == punch_type]


    for start in range(0, len(subset), window_size):
        end = start + window_size
        if end > len(subset):  # Ensure window does not exceed data length
            break
        window = subset.iloc[start:end]

        # Extract accelerometer data
        ax = window['aX'].values
        ay = window['aY'].values
        az = window['aZ'].values
        magnitude = (ax**2 + ay**2 + az**2)**0.5  # Vector magnitude

        features = {
           # 'mean_aX': ax.mean(),
           # 'mean_aY': ay.mean(),
           # 'mean_aZ': az.mean(),
           # 'std_aX': ax.std(),
           # 'std_aY': ay.std(),
           # 'std_aZ': az.std(),
            'max_magnitude': magnitude.max(),
           # 'min_magnitude': magnitude.min(),
            'energy': (magnitude**2).sum(),
            'label': punch_type
        }
        features_list.append(features)
```

```python
features_df = pd.DataFrame(features_list)
features_df.to_csv("Full_features.csv", index=False)
print("Feature extraction completed. Features saved to 'Full_features.csv'.")
```
